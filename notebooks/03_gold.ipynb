```python
import pyspark.sql.functions as F
from pyspark.sql.window import Window

df_vendas = spark.read.table("projeto_silver.vendas_tratadas")
```

```python
display(df_vendas)
```

```python
dim_filial = df_vendas.select(
    "filial",
    "cidade"
).distinct()
```

```python
window = Window.orderBy("filial")

dim_filial =  dim_filial.withColumn(
    "id_filial", 
    F.row_number().over(window)
)

dim_filial = dim_filial.select(
    "id_filial",
    "filial",
    "cidade"
)


display(dim_filial)
```

```python
dim_categoria = df_vendas.select(
    "linha_produto",
).distinct()

```

```python
window = Window.orderBy("linha_produto")

dim_categoria = dim_categoria\
    .withColumn("id_categoria", F.row_number().over(window))

    
```

```python
dim_categoria = dim_categoria.select(
    "id_categoria",
    "linha_produto"
)

display(dim_categoria)
```

```python
dim_calendario = df_vendas.select(
    "data_venda"
).distinct()

display(dim_calendario)
```

```python
dim_calendario = dim_calendario \
    .withColumn("ano", F.year("data_venda")) \
    .withColumn("mes", F.month("data_venda")) \
    .withColumn("trimestre", F.quarter("data_venda")) \
    .withColumn("dia_mes", F.dayofmonth("data_venda")) \
    .withColumn("dia_semana", F.date_format("data_venda", "EEEE")) \
    .withColumn("dia_semana_numero", F.dayofweek("data_venda"))
```

```python
display(dim_calendario)
```

```python
df_vendas = (
    df_vendas
    .join(
    dim_filial,
    on=["filial", "cidade"],
    how="left")
    .join(
    dim_categoria,
    on= ["linha_produto"],
    how = "left"
))

```

```python
display(df_vendas)
```

```python
fato_vendas = df_vendas.select(
    "id_nota_fiscal",
    "id_filial",
    "id_categoria",
    "tipo_cliente",
    "genero",
    "preco_unitario",
    "quantidade",
    "imposto_5_porcento",
    "valor_venda",
    "data_venda",
    "hora_venda",
    "forma_pagamento",
    "custo_produtos_vendidos",
    "percentual_margem_bruta",
    "lucro_bruto",
    "avaliacao"
)

display(fato_vendas)
```

```python
spark.sql("CREATE DATABASE IF NOT EXISTS projeto_gold")

dim_filial.write.format("delta") \
    .mode('overwrite') \
    .saveAsTable("projeto_gold.dim_filial")

dim_categoria.write.format("delta") \
    .mode('overwrite') \
    .saveAsTable("projeto_gold.dim_categoria")

dim_calendario.write.format("delta") \
    .mode('overwrite') \
    .saveAsTable("projeto_gold.dim_calendario")

fato_vendas.write.format("delta") \
    .mode('overwrite') \
    .saveAsTable("projeto_gold.fato_vendas")
```
